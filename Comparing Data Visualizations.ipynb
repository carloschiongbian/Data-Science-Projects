# Comparing Data Visualizations
This notebook compares life expectancy in Asia and Africa in 1952 using the `countries.csv` dataset.

```python
# Import pandas to load and work with the CSV data.
# Import matplotlib.pyplot to create the visualizations.
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the countries.csv file into a pandas DataFrame.
# Keep countries.csv in the same folder as this notebook, or update the path below.
countries = pd.read_csv('countries.csv')

# Display a few rows to check that the data loaded correctly.
countries.head()
```

```python
# Filter the dataset to the year 1952 and the two regions being compared.
# This makes sure the visualization compares the same year for both continents.
data_1952 = countries[countries['year'] == 1952]
asia = data_1952[data_1952['continent'] == 'Asia']
africa = data_1952[data_1952['continent'] == 'Africa']

# Create a time-series-style line plot of life expectancy by country.
# Since only 1952 is requested, countries are used on the x-axis and life expectancy on the y-axis.
# Sort values so the lines are easier to read.
asia = asia.sort_values('lifeExp')
africa = africa.sort_values('lifeExp')

plt.figure(figsize=(12, 5))
plt.plot(asia['country'], asia['lifeExp'], marker='o', label='Asia')
plt.plot(africa['country'], africa['lifeExp'], marker='o', label='Africa')
plt.title('Life Expectancy by Country in 1952: Asia vs Africa')
plt.xlabel('Countries')
plt.ylabel('Life expectancy (years)')
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Create two side-by-side subplots to compare life expectancy distributions.
# hist() groups life expectancy values into bins, making the spread of each continent easy to compare.
plt.figure(figsize=(12, 5))

# Histogram for Asia.
plt.subplot(1, 2, 1)
plt.hist(asia['lifeExp'], bins=10, edgecolor='black')
plt.title('Asia: Life Expectancy in 1952')
plt.xlabel('Life expectancy (years)')
plt.ylabel('Number of countries')

# Histogram for Africa.
plt.subplot(1, 2, 2)
plt.hist(africa['lifeExp'], bins=10, edgecolor='black')
plt.title('Africa: Life Expectancy in 1952')
plt.xlabel('Life expectancy (years)')
plt.ylabel('Number of countries')

# Adjust spacing so the titles and labels do not overlap, then display both histograms.
plt.tight_layout()
plt.show()
```
